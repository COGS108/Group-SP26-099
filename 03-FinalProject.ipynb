## Abstract

This study investigates whether critic scores, user scores, publisher size, and multi-platform release strategy predict total video game sales for action games, and whether those same relationships hold for adventure games released between 2001 and 2020. A linear regression model trained on action games explained 41.6% of the variance in action game sales (R² = 0.416), with platform release count (+0.229) and critic scores (+0.164) emerging as the strongest positive predictors, while publisher ranking had a negligible positive effect and user scores contributed negatively. The model was then applied to adventure games without retraining, where it retained about 80% of its explanatory power but accuracy dropped, with R² falling to 0.334 and prediction error nearly doubling from 0.713 to 1.190 RMSE.

When the model was retrained directly on adventure games, it recovered most of the lost accuracy (R² = 0.430), but more importantly revealed two relationships that flipped direction cross-genre. Critic scores, which positively predicted action game sales (+0.164), became slightly negative for adventure games (−0.011), while user scores flipped from slightly negative in action (−0.030) to a meaningful positive predictor in adventure (+0.152) — a 601.9% change in magnitude. Publisher size and platform count remained consistent positive predictors in both genres, with platform count showing an even stronger effect in adventure (+0.356 vs +0.229).

These findings suggest that action and adventure game buyers behave differently. Action game buyers are influenced by critical acclaim while adventure buyers are more driven by community word-of-mouth. More importantly, a single model cannot fully capture both genres without accounting for these structural differences.

Overall, the exploratory analysis indicates that video game sales are influenced by a combination of factors rather than any single characteristic. The findings highlight the importance of examining how review scores, publisher characteristics, and platform availability interact with one another to better understand what contributes to higher sales performance.

### Experimental Analysis (part 1): What Predicts Total Sales for Action Games?
### Standard ordinary least squares (OLS) linear regression

The first part of our research question asks whether critic scores, user scores, publisher portfolio size, and multi-platform releases predict higher total sales for Action games. To test this, we fit a linear regression model using only games labeled as `Action`. The outcome variable is `log_total_sales_per_title`, which represents total title-level sales after a log transformation. We use the log-transformed version because the EDA showed that raw sales are strongly right-skewed. The predictors are `publisher_rank`, `platform_release_count`, `metacritic_review_score`, and `user_review_score`.
Publisher rank must be interpreted differently because lower rank values represent larger publishers. Therefore, a negative coefficient for `publisher_rank` means that games from larger publishers are associated with higher predicted sales.

For this first analysis, we use the following process:

1. Filter the dataset to Action games.
2. Define the predictors and target variable.
3. Fit a linear regression model to predict log-transformed title-level sales.
4. Evaluate the model using R² and RMSE.
5. Interpret the regression coefficients and predicted-vs-actual plot.

```python
# Splitting data for the Action genre
action = games[games['genre'] == 'Action'].copy()

# Defining predictors and target
features = ['publisher_rank', 'platform_release_count',
            'metacritic_review_score', 'user_review_score']

target = 'log_total_sales_per_title'

X_action = action[features]
y_action = action[target]

# Fitting linear regression model on Action games
action_model = LinearRegression()
action_model.fit(X_action, y_action)

# Predicting Action game sales
y_action_pred = action_model.predict(X_action)

# Evaluating model fit
action_r2 = r2_score(y_action, y_action_pred)
action_rmse = np.sqrt(mean_squared_error(y_action, y_action_pred))
action_mae = mean_absolute_error(y_action, y_action_pred)

print("EValuating model fit on Action games")
print(f"R²:   {action_r2:.3f}")
print(f"RMSE: {action_rmse:.3f}")
print(f"MAE:  {action_mae:.3f}")

print("\nCoefficients:")
for feat, coef in zip(features, action_model.coef_):
    print(f"  {feat}: {coef:.4f}")
print(f"  Intercept: {action_model.intercept_:.4f}")

plt.figure(figsize=(8, 6))

sns.scatterplot(x=y_action, y=y_action_pred, alpha=0.5)
plt.plot([y_action.min(), y_action.max()], [y_action.min(), y_action.max()],
         linestyle="--", color="black")

plt.title("Action Games: Predicted vs Actual Log Sales", fontsize=14, pad=12)
plt.xlabel("Actual Log10(Total Sales per Title + 1)", fontsize=12)
plt.ylabel("Predicted Log10(Total Sales per Title + 1)", fontsize=12)
plt.tight_layout()
plt.show()
```

The Action game regression model explains a moderate amount of variation in title-level sales. The R² value is 0.416, meaning that publisher rank, platform release count, critic score, and user score together explain about 41.6% of the variation in log-transformed total sales for Action games. The RMSE is 0.713, which means the model’s typical prediction error is about 0.713 log-sales units.

The coefficients suggest that platform availability and critic score are positively associated with Action game sales. `platform_release_count` has a positive coefficient of 0.2289, meaning that Action games released on more platforms tend to have higher predicted total sales. `metacritic_review_score` also has a positive coefficient of 0.1637, suggesting that higher critic scores are associated with higher predicted sales for Action games.

`publisher_rank` has a negative coefficient of -0.0037. Since lower publisher rank values represent larger publishers, this means that games from larger publishers tend to have higher predicted sales. In contrast, `user_review_score` has a slightly negative coefficient of -0.0303 after controlling for the other predictors, suggesting that user scores do not add positive predictive value for Action game sales in this model.

The predicted-vs-actual plot shows that the model captures some broad sales patterns, especially for games with moderate to high sales. However, the points are still spread around the diagonal reference line, meaning the model does not fully explain Action game sales. The vertical stacks near zero actual sales also show that the model tends to overpredict some games with very low or zero recorded sales. This suggests that other factors not included in the model, such as franchise popularity, marketing budget, platform exclusivity, and release timing, may also play an important role.

Overall, Analysis 1 provides partial support for the first part of the research question. For Action games, higher critic scores, larger publisher portfolio size, and more platform releases are associated with higher predicted sales. However, user scores do not show the same positive relationship in this model. Because the model explains only part of the variation in sales, the next analysis tests whether these same relationships transfer to Adventure games or whether genre-specific patterns affect prediction accuracy.

### Experimental Analysis Part 2: Can the Action Game Sales Model Predict Adventure Game Sales?

The second part of our research question asks whether the sales relationships learned from Action games can accurately predict total sales for Adventure games. In Analysis Part 1, we trained an ordinary least squares linear regression model on Action games using publisher rank, platform release count, Metacritic critic score, and Metacritic user score to predict `log_total_sales_per_title`.

In this second analysis, we apply that Action-trained model to the Adventure game subset without retraining it. This lets us test whether the same relationships between publisher size, platform availability, review scores, and sales transfer across genres. We evaluate the transfer model using R², RMSE, and MAE, then compare it to a separate model retrained directly on Adventure games. If the retrained Adventure model performs better or learns different coefficients, that suggests the sales patterns are not identical across the two genres.

For this second analysis, we use the following process:

1. Apply the Action-trained regression model to Adventure games.
2. Evaluate cross-genre prediction accuracy using R², RMSE, and MAE.
3. Compare predicted Adventure sales against actual Adventure sales using a predicted-vs-actual plot.
4. Retrain the same regression model directly on Adventure games.
5. Compare Action and Adventure model performance and coefficients to see whether the predictors behave similarly across genres.

```python
# Splitting data for the Adventure genre
adventure = games[games['genre'] == 'Adventure'].copy()

X_adventure = adventure[features]
y_adventure_empirical = adventure[target]

print("Number of Action games:", len(action))
print("Number of Adventure games:", len(adventure))

# Predict Adventure sales using the Action-trained model
y_adventure_predicted = action_model.predict(X_adventure)

# Evaluate cross-genre prediction accuracy
adventure_transfer_r2 = r2_score(y_adventure_empirical, y_adventure_predicted)
adventure_transfer_rmse = np.sqrt(mean_squared_error(y_adventure_empirical, y_adventure_predicted))
adventure_transfer_mae = mean_absolute_error(y_adventure_empirical, y_adventure_predicted)

print("Cross-genre accuracy: Action model onto Adventure games")
print(f"R²:   {adventure_transfer_r2:.3f}")
print(f"RMSE: {adventure_transfer_rmse:.3f}")
print(f"MAE:  {adventure_transfer_mae:.3f}")

# Predicted vs actual Adventure plot
plt.figure(figsize=(8, 6))

sns.scatterplot(x=y_adventure_empirical, y=y_adventure_predicted, alpha=0.5)
plt.plot([y_adventure_empirical.min(), y_adventure_empirical.max()], [y_adventure_empirical.min(), y_adventure_empirical.max()],
         linestyle="--", color="black")
plt.title("Adventure Games: Predicted vs Actual Log Sales Using Action Model", fontsize=14, pad=12)
plt.xlabel("Actual Log10(Total Sales per Title + 1)", fontsize=12)
plt.ylabel("Predicted Log10(Total Sales per Title + 1)", fontsize=12)
plt.tight_layout()
plt.show()
```

The predicted-vs-actual plot shows that the Action-trained model captures some of the general pattern in Adventure game sales, but it does not predict them perfectly. Many points are near the diagonal reference line, especially for games with moderate-to-high actual sales, which suggests that the Action model is not useless when applied to Adventure games.

However, the model also shows clear errors. Some Adventure games with zero or very low recorded sales are predicted to have much higher sales, creating vertical stacks on the left side of the plot. The model also compresses predictions toward the middle range, meaning it does not fully capture the lowest- and highest-selling Adventure games. This supports the idea that Adventure games may have sales patterns that are not fully captured by a model trained only on Action games.

## Discusison and Conclusion

This project investigated whether critic scores, user scores, publisher portfolio size, and platform availability could predict total sales for Action games and whether those relationships transfer to Adventure games. Our analysis found that platform availability and publisher portfolio size were among the strongest factors related to sales. Games released on multiple platforms generally sold more copies, while games published by companies with larger portfolios tended to perform better overall. Critic scores also showed a connection to sales, although user scores appeared to be a less consistent factor. These findings partially support our original hypothesis that review scores and publisher-related factors would be associated with a game's sales. In the Action-game regression model, the four predictors explained about 41.6% of the variation in log-transformed title-level sales. When the Action-trained model was applied to Adventure games, performance dropped to an R² of 0.334. However, when the model was retrained directly on Adventure games, the R² increased to 0.430. This suggests that the same variables are useful across both genres, but the relationships do not transfer perfectly from Action games to Adventure games.

The exploratory data analysis revealed that video game sales are highly skewed, with a small number of blockbuster titles accounting for a large portion of total sales. This pattern suggests that commercial success in the gaming industry is concentrated among relatively few games. While many titles received similar review scores, their sales varied dramatically, suggesting that factors beyond review score alone are associated with financial success. This finding highlights the importance of considering both product characteristics and industry-level factors when studying game sales.

One of the most interesting results emerged when comparing action and adventure games. For action games, critic scores appeared to be more closely related to sales performance, while user scores became more influential for adventure games. This suggests that different player communities may rely on different sources of information when making purchasing decisions. In contrast, publisher size and platform availability remained important across genres, reinforcing the idea that accessibility and publisher resources have broad effects on commercial performance. The coefficient comparison supports this genre difference. In the Action model, critic score had a positive coefficient, while user score had a slightly negative coefficient after controlling for publisher rank and platform release count. In the retrained Adventure model, the critic score was close to zero and slightly negative, while the user score became positive. In contrast, publisher rank and platform release count kept the same direction across both genres. This suggests that publisher size and platform availability are more stable predictors across genres, while review scores behave differently depending on genre.

These results extend the ideas discussed in our background research by showing that commercial success cannot be explained by a single variable. Instead, sales appear to result from a combination of visibility, accessibility, publisher influence, and perceived quality. The fact that review scores behaved differently across genres also demonstrates that consumer behaviour is more complex than a simple "higher scores equal higher sales" relationship.

There are several limitations to this analysis. The dataset only includes games released between 2001 and 2020 and does not account for factors such as marketing budgets, franchise popularity, release timing, or social media engagement. Publisher rank is also only a proxy for publisher portfolio size, and may reflect other factors such as brand recognition, marketing power, distribution reach, or franchise strength. In addition, the data are observational, meaning we can identify associations but cannot determine causation.

The dataset itself also shapes what we can conclude. Because the analysis relies on recorded copy sales, it underrepresents free-to-play games and games whose success is measured through downloads, active users, subscriptions, or in-game purchases. The dataset may also favor games with enough visibility to have Metacritic coverage and reported sales, meaning smaller or less commercially visible games may be less represented. Future work could incorporate additional variables and examine more genres to better understand the factors associated with video game success. Overall, our findings suggest that while review scores matter, publisher resources and market reach are more consistent predictors of commercial performance across Action and Adventure games.
