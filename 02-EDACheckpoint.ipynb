# Research Question

Does increasing Metacritic critic scores, Metacritic user scores, publisher portfolio size, and multi-platform releases predict higher total sales for action games, and can the same relationships accurately predict total sales for adventure games released between 2001 and 2020?

## Background and Prior Work

In 2025, the gaming industry reached a record-breaking $190 billion in market size. Gaming is widespread across the world, with players spanning multiple countries and continents, demonstrating that video games are a shared but culturally influenced activity. Among all genres, action games consistently rank as one of the most popular worldwide, making them a useful focus for analyzing player preferences and consumer behavior.

Previous research has explored how review scores influence video game sales performance. A 2024 study found that higher user review scores were associated with a decrease in total video game sales, while higher critic review scores showed an exponential increase in sales.<a name="ref1"></a><sup>1</sup> These findings suggest that professional critical reception may have a stronger impact on commercial success than player reception alone, highlighting the importance of critic influence and reputation within the gaming market.

A 2021 study found that action games consistently generate the highest global sales revenues across all genres, while North America leads the world in consumer spending on video games.<a name="ref1"></a><sup>2</sup> These trends reinforce the importance of studying action games specifically, as they dominate the market and attract broad international audiences. 

Most recently, a 2026 study found that certain publishers, most notably Nintendo, consistently produce higher-selling games. This suggests that publisher size, name-recognition and reputation contribute heavily to each title's commercial success.<a name="ref1"></a><sup>3</sup> However, the study concluded that no single factor alone can fully explain a game’s performance, emphasizing the need to analyze multiple attributes together.

Building on this prior work, we aim to examine how factors such as Metacritic critic scores, Metacritic user scores, publisher portfolio size, and multi-platform releases relate to and predict total sales for action games released between 2001 and 2020. Metacritic started in 2001 so we will not include any titles released before then. We chose to focus on action games because they consistently dominate global sales revenues and provide a strong basis for analyzing how critical reception, publisher influence, and platform availability contribute to commercial success. Additionally, we seek to determine whether these same relationships can be used to predict total sales for adventure games, allowing us to understand market differences between action and adventure genres.

```python
# Read in the csv to a dataframe
game_sales_raw = pd.read_csv('data/00-raw/video_games_sales_1980_2024.csv')
print(game_sales_raw.columns)

# Get only necessary columns
game_sales = game_sales_raw[['title','console','genre','publisher','critic_score', 'total_sales', 'release_year']].copy()
game_sales = game_sales.dropna() # drop rows with null values

# Make release_year integers
game_sales['release_year'] = game_sales['release_year'].astype(int)

# let's only focus on action and adventure games
print("Unique List of game genres:", game_sales['genre'].unique())
game_sales_action_adventure = game_sales[game_sales['genre'].isin(['Action', 'Adventure', 'Action-Adventure'])] # allow action-adventure games to be included

# Preview Dataset 1 (Action)
print("Number of rows in dataset 1 is:", game_sales_action_adventure.shape[0])
game_sales_action_adventure.head()
```

### Dataset #2
 as above, add any more copies of this that you need to given how many datasets you have

```python
# Read in the csv to a dataframe
critic_user_scores_raw = pd.read_csv('data/00-raw/metacritic_user_critic_scores_2024.csv')
print(critic_user_scores_raw.columns)

# Get only necessary columns
critic_user_scores = critic_user_scores_raw[['name','metacritic_review_count','metacritic_review_score','user_review_count','user_review_score','genres']].copy()
critic_user_scores = critic_user_scores.dropna() # drop rows with null values
print("Unique sub-genres here are", critic_user_scores['genres'].unique())

# Make the metacritic_review_score out of 10 instead of 100
critic_user_scores['metacritic_review_score'] = (pd.to_numeric(critic_user_scores['metacritic_review_score'], errors='coerce') / 10)

# Remove "Based on xxx Critic/User Ratings" string and just get the numbers as integers
def extract_review_count(i):
    cleaned = ''.join(char for char in str(i) if char.isdigit() or char == ',')
    return int(cleaned.replace(',', '')) if cleaned else 0
critic_user_scores['metacritic_review_count'] = critic_user_scores['metacritic_review_count'].apply(extract_review_count)
critic_user_scores['user_review_count'] = critic_user_scores['user_review_count'].apply(extract_review_count)

# Remove the ['xxx'] part of the string
critic_user_scores['genres'] = critic_user_scores['genres'].astype(str).str.strip("[]").str.replace("'", "", regex=False)
critic_user_scores = critic_user_scores.rename(columns={'genres': 'sub-genre'})

# Preview Dataset 2
print("Number of rows in dataset 2 is:", critic_user_scores.shape[0])
critic_user_scores.head()
```

### Merge Both Datasets for Analysis

```python
# Merge both datasets and keep only games that match the titles and names in both datasets
# Without sales numbers, there's no point in trying to keep the ones that are in dataset #2 but not in dataset #1
game_merged_df = game_sales_action_adventure.merge(critic_user_scores, left_on = 'title', right_on = 'name', how='inner')
game_merged_df = game_merged_df[game_merged_df['release_year'] >= 2001]
print("Number of rows in merged dataset is:", game_merged_df.shape[0])

# There are mismatchings betweent the critic_score from dataset #1 and the ones in dataset #2. 
# I'm choosing to only keep scores from dataset #2 because it shows user scores too.
game_ultimate_df = game_merged_df[['title','console','genre','publisher','metacritic_review_count','metacritic_review_score','user_review_count','user_review_score','total_sales','release_year']].copy()

# Publisher size is based on total sales from dataset #1
game_sales_raw_publisher = game_sales_raw[['publisher','total_sales']].copy()
game_sales_raw_publisher['total_sales'] = (game_sales_raw_publisher['total_sales'] * 10**6).astype(int)
game_sales_raw_publisher = game_sales_raw_publisher.groupby('publisher')['total_sales'].sum().reset_index().sort_values(by='total_sales', ascending=False)
print('The publisher with the highest revenue as of 2020 is', game_sales_raw_publisher['publisher'].iloc[0], 'at', game_sales_raw_publisher['total_sales'].iloc[0], 'in total sales.')

# Create a rank column for publishers. Highest total sales = 1 and the number increases the lower down you go.
game_sales_raw_publisher['publisher_rank'] = range(1, len(game_sales_raw_publisher) + 1)

# Merge this publisher_rank column back to game_ultimate_df
game_sales_raw_publisher = game_sales_raw_publisher[['publisher','publisher_rank']].copy()
game_ultimate_df = game_ultimate_df.merge(game_sales_raw_publisher, left_on='publisher', right_on='publisher', how='left')

# Create Statistics Columns
game_ultimate_df['platform_release_count'] = (game_ultimate_df.groupby('title')['title'].transform('size')) # count number of platform releases per title
game_ultimate_df['total_sales'] = (game_ultimate_df['total_sales'] * 10**6).astype(int) # change total sales to be in the millions
game_ultimate_df['total_sales_per_title'] = (game_ultimate_df.groupby('title')['total_sales'].transform('sum')) # total sales across all platforms per title

game_ultimate_df.head(10)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Loading wrangled dataset from data/02-processed
games = pd.read_csv("data/02-processed/game_ultimate_data.csv")
```

The processed dataset contains 1,943 rows and 13 columns. Each row represents a game-console release rather than a unique game title, so the same game can appear multiple times if it was released across multiple platforms. For example, *Grand Theft Auto V* appears on PS3, PS4, Xbox 360, and Xbox One.

This structure matters because `total_sales` represents sales for a specific game-console release, while `total_sales_per_title` aggregates sales across all platform releases for the same game title. Since our research question focuses on how sales relate to review scores, publisher size, platform availability, and genre, both platform-level and title-level sales may be useful.

The dataset includes four categorical variables: `title`, `console`, `genre`, and `publisher`. The remaining variables are numeric and include review counts, review scores, sales, release year, publisher rank, platform release count, and total sales per title. All columns have 1,943 non-null values, so there are no missing values visible from the basic dataset structure check.

Next, we summarize the main numeric variables that are relevant to our research question. This helps us understand the typical values, ranges, and variability in review scores, sales, publisher rank, release year, and platform release count before making plots or comparing relationships between variables.

```python
numeric_vars = [
    "metacritic_review_count",
    "metacritic_review_score",
    "user_review_count",
    "user_review_score",
    "total_sales",
    "release_year",
    "publisher_rank",
    "platform_release_count",
    "total_sales_per_title"
]

games[numeric_vars].describe().T
```

The summary statistics show that the review score variables fall within the expected 0–10 range. Metacritic critic scores range from 2.3 to 9.8, with a median score of 7.0. User scores range from 1.0 to 9.3, with a median score of 7.3. This suggests that most games in the dataset are rated around the middle-to-high range, but there are still some poorly rated games on both critic and user scales.

The sales variables appear much more uneven. The median platform-level `total_sales` is 180,000 copies, but the mean is about 511,168 copies and the maximum is 20,320,000 copies. This large gap between the median, mean, and maximum suggests that sales are strongly right-skewed: most game-console releases sell relatively modest numbers, while a small number of major games sell extremely high numbers of copies.

The same pattern appears for `total_sales_per_title`. The median title-level sales value is 600,000 copies, while the mean is about 1.7 million copies and the maximum is 64.29 million copies. This makes sense because `total_sales_per_title` combines sales across all platform releases for the same game title.

The review count variables are also highly uneven. For example, the median user review count is 59, while the maximum is 28,071. This suggests that some major games receive far more user engagement than most games in the dataset. Because sales and review counts are skewed, later plots should avoid relying only on raw means and should use distributions, medians, boxplots, and log-transformed sales where appropriate.

Because both sales variables are strongly right-skewed, we create log-transformed versions of platform-level and title-level sales. This does not replace the original sales variables, but it makes the distributions and later scatterplots easier to interpret by reducing the visual influence of extremely high-selling games.

```python
# Creating log-transformed sales variables to make skewed sales distributions easier to visualize
games["log_total_sales"] = np.log10(games["total_sales"] + 1)
games["log_total_sales_per_title"] = np.log10(games["total_sales_per_title"] + 1)

games[["total_sales", "log_total_sales", "total_sales_per_title", "log_total_sales_per_title"]].head()
```

To better understand the shape of the sales variables, we first compare the raw `total_sales` distribution with its log-transformed version. The summary statistics suggested that sales are highly right-skewed, so these plots help confirm whether a small number of very high-selling games are visually dominating the distribution.

```python
sns.set_theme(style="whitegrid", context="notebook")
fig, axes = plt.subplots(1, 2, figsize=(18, 6))

# Raw total sales distribution
sns.histplot(data=games, x="total_sales", bins=40, ax=axes[0])
axes[0].set_title("Distribution of Platform-Level Total Sales", fontsize=14, pad=12)
axes[0].set_xlabel("Total Sales", fontsize=12)
axes[0].set_ylabel("Number of Game-Console Releases", fontsize=12)
axes[0].tick_params(axis="both", labelsize=10)

# Log-transformed total sales distribution
sns.histplot(data=games, x="log_total_sales", bins=40, kde=True, ax=axes[1])
axes[1].set_title("Distribution of Log-Transformed Platform-Level Sales", fontsize=14, pad=12)
axes[1].set_xlabel("Log10(Total Sales + 1)", fontsize=12)
axes[1].set_ylabel("Number of Game-Console Releases", fontsize=12)
axes[1].tick_params(axis="both", labelsize=10)

plt.subplots_adjust(wspace=0.25)
plt.show()
```

The raw `total_sales` distribution is strongly right-skewed. Most game-console releases have relatively low sales, while a small number of major releases sell millions of copies. This matches the summary statistics, where the median platform-level sales value was 180,000 copies but the maximum was over 20 million copies.

The log-transformed sales plot is easier to interpret because it reduces the visual effect of extremely high-selling games. After the transformation, the distribution of most nonzero-selling games becomes much more visible instead of being compressed near zero. Because of this skew, later relationship plots will be using `log_total_sales` when comparing sales to review scores, publisher rank, and platform release count.

```python
zero_sales_count = (games["total_sales"] == 0).sum()
zero_sales_percent = round((zero_sales_count / len(games)) * 100, 2)

print("Number of rows with zero total sales:", zero_sales_count)
print("Percent of rows with zero total sales:", zero_sales_percent)
```

This variable is relevant because games released on more platforms may have more opportunities to generate sales than games released on only one console.

Most game titles in the dataset have relatively few platform releases. The largest groups are games with 1, 2, or 3 platform releases, while games released on 6, 7, or 8 platforms are much less common. This shows that platform availability is uneven across games.

This matters for the project because games released on more platforms may have more opportunities to accumulate sales. Because of this, `platform_release_count` will be considered when interpreting `total_sales_per_title`, especially when comparing games that were only released on one console to games that were released more widely.

Overall, this first part of the EDA shows that the processed dataset is ready for further analysis. The dataset has 1,943 complete game-console release observations with no missing values in the variables used for analysis. Review scores are mostly concentrated in the middle-to-high range, while sales are strongly right-skewed, meaning a small number of games account for very large sales totals.

Because of this skew, our later EDA will use log-transformed sales when examining relationships between sales and review scores, publisher rank, and platform release count. The platform release count distribution also shows that games differ in how widely they were released, which means platform availability may be an important factor when interpreting total title-level sales.

#### Section 2 of EDA: Relationships Between Reviews, Genre, Publisher Size, Platform Availability, and Sales

For Section 2, we start exploring relationships between the main variables in our research question. 

```python
critic_sales = games[games["log_total_sales"] > 0]

plt.figure(figsize=(10, 6))

sns.regplot(
    data=critic_sales,
    x="metacritic_review_score",
    y="log_total_sales",
    scatter_kws={"alpha": 0.4},
    line_kws={"color": "red"}
)

plt.title("Critic Scores vs Log Total Sales")
plt.xlabel("Metacritic Critic Score")
plt.ylabel("Log Total Sales")
plt.show()
```

The scatterplot shows a positive relationship between critic review scores and game sales. Games with higher Metacritic critic scores generally tend to have higher total sales, supporting our hypothesis that stronger critical reception is associated with greater commercial success for video games, particularly within the action genre.

```python
plt.figure(figsize=(10, 6))

sns.regplot(
    data=games,
    x="user_review_score",
    y="log_total_sales",
    scatter_kws={"alpha": 0.4},
    line_kws={"color": "red"}
)

plt.title("User Scores vs Log Total Sales")
plt.xlabel("User Review Score")
plt.ylabel("Log Total Sales")
plt.show()
```

User review scores also show a positive association with total sales, though the relationship appears weaker than critic scores. This suggests that player reception may still contribute to commercial success, but professional critic reviews may serve as a stronger predictor of sales.

```python
plt.figure(figsize=(10, 6))

sns.regplot(
    data=games,
    x="metacritic_review_score",
    y="user_review_score",
    scatter_kws={"alpha": 0.4},
    line_kws={"color": "red"}
)

plt.title("Critic Scores vs User Scores")
plt.xlabel("Metacritic Critic Score")
plt.ylabel("User Review Score")
plt.show()
```

Critic scores and user scores display a moderately positive relationship, suggesting that critics and players often agree on overall game quality. However, the spread in the data indicates that disagreement still exists for some titles. Certain games may receive strong critic reviews but mixed user reactions, while others develop strong fan support despite weaker critic reception. This demonstrates that critics and consumers evaluate games using somewhat different standards and expectations.

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=games,
    x="platform_release_count",
    y="log_total_sales"
)

plt.title("Platform Release Count vs Log Total Sales")
plt.xlabel("Number of Platforms Released On")
plt.ylabel("Log Total Sales")
plt.show()
```

Games released on more platforms generally show higher sales values and wider sales distributions than games released on only one platform. Multi-platform availability likely increases market reach by exposing games to larger audiences across different consoles and systems. However, some variability still exists because not every multi-platform release becomes commercially successful. Additionally, highly successful games may later be ported to additional platforms after their initial success.

```python
plt.figure(figsize=(10, 6))

sns.regplot(
    data=games,
    x="publisher_rank",
    y="log_total_sales",
    scatter_kws={"alpha": 0.4},
    line_kws={"color": "red"}
)

plt.title("Publisher Rank vs Log Total Sales")
plt.xlabel("Publisher Rank")
plt.ylabel("Log Total Sales")
plt.show()
```

The relationship between publisher rank and sales suggests that publisher prominence may influence commercial success. Games published by more prominent publishers tend to achieve higher sales overall. Larger or more established publishers may benefit from greater marketing resources, broader distribution networks, and stronger franchise recognition. However, variability in the data shows that publisher rank alone does not fully determine a game's commercial performance.

```python
genre_sales = (
    games.groupby("genre")["total_sales"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=genre_sales.index,
    y=genre_sales.values
)

plt.title("Average Total Sales by Genre")
plt.xlabel("Genre")
plt.ylabel("Average Total Sales")
plt.xticks(rotation=45)
plt.show()
```

Average sales vary noticeably across genres. Some genres consistently produce higher-selling games, while others tend to have lower commercial performance. Action games appear to have some of the highest average sales among genres in the dataset, which supports the project hypothesis that action-oriented titles may have greater market appeal. Differences between genres may reflect variations in audience size, replayability, accessibility, and franchise popularity.

```python
plt.figure(figsize=(12, 6))

sns.boxplot(
    data=games,
    x="genre",
    y="metacritic_review_score"
)

plt.title("Critic Review Scores by Genre")
plt.xlabel("Genre")
plt.ylabel("Metacritic Critic Score")
plt.xticks(rotation=45)
plt.show()
```

Review score distributions differ across genres, suggesting that critics evaluate certain genres more favorably than others. Genres such as Action and Role-Playing show higher median critic scores and smaller interquartile ranges than some other genres, suggesting that these genres tend to receive more consistently positive critic reviews. This may reflect differences in gameplay expectations, storytelling quality, innovation, or technical performance across genres.

```python
yearly_sales = (
    games.groupby("release_year")["total_sales"]
    .mean()
)

plt.figure(figsize=(12, 6))

sns.lineplot(
    x=yearly_sales.index,
    y=yearly_sales.values,
    marker="o"
)

plt.title("Average Total Sales by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Average Total Sales")
plt.show()
```

Average total sales fluctuated across release years, with a significant peak around 2013 followed by a general decline in later years. This suggests that certain release periods were considerably more successful than others, possibly due to popular game releases or changing market trends.

Overall, the trends observed throughout Section 2 support the project hypothesis. Higher review scores, stronger publisher rankings, and broader platform availability all appear positively associated with stronger sales performance, particularly among action-oriented titles.
